```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import zipfile
import os

# Define the path to the provided zip file
zip_file_path = '/content/archive.zip'

# Define the directory to extract files to
extract_dir = '/content/extracted_data'
os.makedirs(extract_dir, exist_ok=True)

# Unzip the archive
with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
    zip_ref.extractall(extract_dir)
print(f"'{zip_file_path}' unzipped to '{extract_dir}'")

# Check the contents of the extracted directory (optional, for debugging)
print("Contents of extracted directory:")
print(os.listdir(extract_dir))

# Assuming 'Mall_Customers.csv' is directly inside the zip file and extracted to extract_dir
data_file_path = os.path.join(extract_dir, 'Mall_Customers.csv')

# Load the dataset
try:
    df = pd.read_csv(data_file_path)
    print(f"File '{data_file_path}' loaded successfully!")
except FileNotFoundError:
    print(f"Error: '{data_file_path}' not found in the extracted archive. Please check the filename.")
    df = None
except Exception as e:
    print(f"An error occurred while loading the CSV: {e}")
    df = None

# Proceed only if the DataFrame was loaded successfully
if df is not None:
    # Display the first few rows
    print("First 5 rows of the dataset:")
    print(df.head())

    # Dataset info
    print("\nDataset Info:")
    print(df.info())

    # Gender Distribution - Pie Chart
    gender_counts = df['Gender'].value_counts()
    plt.figure(figsize=(6, 6))
    plt.pie(gender_counts, labels=gender_counts.index, autopct='%1.1f%%', startangle=90, colors=['skyblue', 'lightgreen'])
    plt.title('Gender Distribution')
    plt.axis('equal')
    plt.show()

    # Age Distribution - Histogram
    plt.figure(figsize=(8, 5))
    sns.histplot(df['Age'], bins=20, kde=True, color='coral')
    plt.title('Age Distribution')
    plt.xlabel('Age')
    plt.ylabel('Frequency')
    plt.show()

    # Annual Income vs Spending Score - Scatter Plot
    plt.figure(figsize=(8, 6))
    sns.scatterplot(data=df, x='Annual Income (k$)', y='Spending Score (1-100)', hue='Gender', palette='Set1')
    plt.title('Annual Income vs Spending Score')
    plt.xlabel('Annual Income (k$)')
    plt.ylabel('Spending Score (1-100)')
    plt.legend(title='Gender')
    plt.show()

    # Pairplot for numerical features
    sns.pairplot(df.drop('CustomerID', axis=1), hue='Gender')
    plt.suptitle('Pairplot of Numerical Features', y=1.02)
    plt.show()
```
